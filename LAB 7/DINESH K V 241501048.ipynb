EXPERIEMNT 1

```
import numpy as np
from scipy.optimize import least_squares
def model(p, x):
  return p[0] * np.exp(p[1] * x)
def residuals(p, x, y):
  return y - model(p, x)
x = np.array([0,1,2,3,4])
y = np.array([1,2.7,7.4,20.1,54.6])
p0 = [1, 0.5]
result = least_squares(residuals, p0, args=(x, y))
print('Optimized Parameters:', result.x)
```
