EXPERIMENT 1

```
def dp_example():
  Weights=[1,3,4,5]
  Values=[1,4,5,7]
  Capacity =7
  n=len(Values)
  dp=[[0 for _ in range(Capacity + 1)] for _ in range(n+1)]
  for i in range(1,n+1):
    for w in range(1,Capacity + 1):
      if Weights[i-1]<=w:
        dp[i][w] = max(
            Values[i-1] + dp[i-1][w-Weights[i-1]],
            dp[i-1][w]
        )
      else:
        dp[i][w] = dp[i-1][w]
  print("Maximum value:", dp[n][Capacity])
dp_example()
```

EXPERIMENT 3

```
def dp_example():
  weights = [2, 3, 4, 5]
  values = [3, 4, 5, 6]
  capacity = 5
  n = len(values)
  dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]
  for i in range(n + 1):
    dp[i][0] = 0
  for w in range(capacity + 1):
    dp[0][w] = 0
  for i in range(1, n + 1):
    for w in range(1, capacity + 1):
      if weights[i - 1] <= w:
        dp[i][w] = max(
          values[i - 1] + dp[i - 1][w - weights[i - 1]],
          dp[i - 1][w]
        )
      else:
          dp[i][w] = dp[i - 1][w]
  result = dp[n][capacity]
  print("Maximum value that can be obtained:", result)
dp_example()
```

```
def dp_example():
  weights = [1, 3, 4, 5]
  values = [1, 4, 5, 7]
  capacity = 7
  n = len(values)
  dp = [[0] * (capacity + 1) for _ in range(n + 1)]
  for i in range(1, n + 1):
    for w in range(1, capacity + 1):
      if weights[i - 1] <= w:
        dp[i][w] = max(values[i - 1] + dp[i - 1][w - weights[i - 1]],
                       dp[i - 1][w])
      else:
        dp[i][w] = dp[i - 1][w]
  print("Result:", dp[n][capacity])
dp_example()
```

EXPERIMENT 6

```
def dp_example():
  coins = [1, 2, 5]
  amount = 11
  dp = [float('inf')] * (amount + 1)
  dp[0] = 0
  for i in range(1, amount + 1):
    for coin in coins:
      if coin <= i:
        dp[i] = min(dp[i], dp[i - coin] + 1)
  result = dp[amount] if dp[amount] != float('inf') else -1
  print("Result:", result)
dp_example()
```

EXPERIMENT 8

```
def dp_example():
  s1 = "AGGTAB"
  s2 = "GXTXAYB"
  m = len(s1)
  n = len(s2)
  dp = [[0] * (n + 1) for _ in range(m + 1)]
  for i in range(1, m + 1):
    for j in range(1, n + 1):
      if s1[i - 1] == s2[j - 1]:
        dp[i][j] = 1 + dp[i - 1][j - 1]
      else:
        dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
  print("Result:", dp[m][n])
dp_example()
```

```
def dp_example():
  arr = [10, 22, 9, 33, 21, 50, 41, 60]
  n = len(arr)
  dp = [1] * n
  for i in range(1, n):
    for j in range(i):
      if arr[i] > arr[j]:
        dp[i] = max(dp[i], dp[j] + 1)
  print("Result:", max(dp))
dp_example()
```

```
def dp_example():
  grid = [
    [1, 3, 1],
    [1, 5, 1],
    [4, 2, 1],
  ]
  m = len(grid)
  n = len(grid[0])
  dp = [[0] * n for _ in range(m)]
  dp[0][0] = grid[0][0]
  for i in range(1, m):
    dp[i][0] = dp[i - 1][0] + grid[i][0]
  for j in range(1, n):
    dp[0][j] = dp[0][j - 1] + grid[0][j]
  for i in range(1, m):
    for j in range(1, n):
      dp[i][j] = grid[i][j] + min(dp[i - 1][j], dp[i][j - 1])
  print("Result:", dp[m - 1][n - 1])
dp_example()
```
