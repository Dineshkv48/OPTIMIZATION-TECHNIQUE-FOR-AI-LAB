EXPERIMENT 1

```
import numpy as np
from scipy.optimize import linprog
cost = [2, 4, 5, 3]
A_eq = [[1, 1, 0, 0], [0, 0, 1, 1]]
b_eq = [20, 30]
result = linprog(cost, A_eq=A_eq, b_eq=b_eq)
print('Minimum Transportation Cost =', result.fun)
```
